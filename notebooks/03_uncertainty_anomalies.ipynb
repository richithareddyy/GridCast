## Interpretation limits

Coverage is empirical under serial dependence and drift. Overlapping condition groups have different sample sizes; 95% coverage overall need not imply 95% coverage during extreme demand. Forecast anomalies are interval violations, not verified physical failures. Monitoring windows use only matured labels; early windows are partial. No test-based retuning is permitted.
